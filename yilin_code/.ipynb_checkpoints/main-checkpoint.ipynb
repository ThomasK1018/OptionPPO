# Vol modeling update

$\begin{align} \sigma_{t}-\sigma_{t-1} & = \alpha (\theta - \sigma_{t-1}) + \beta \sigma_{t-1} Z & \text{ Transitional Equation}\\
\sigma_{t} & = \lambda \ln \frac{H_{t}}{L_{t}}+\epsilon & \text{ Observation Equation}
\end{align}$

There are four variables to work on: $\alpha, \theta, \beta, \lambda$ and two observations $H_{t}, L_{t}$. Now by solving for the four equations, we can estimate $\hat{\sigma_{t}}$, and then use the respective $dS_{t}$ equation to simulate $S_{t}$, and finally calculate $C_{t} / P_{t}$ accordingly.

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Data from the table
pair_labels = ["Top-5", "Top-10"]
metrics = ["Precision", "Normalized Recall", "MRR", "Binary nDCG"]

# Values from the table (each pair: BGE first, trained reranker second)
grouped_values = np.array([
    [0.5913, 0.6043, 0.6570, 0.7540],  
    [0.7324, 0.7456, 0.7078, 0.8124],  

    [0.3771, 0.6881, 0.3854, 0.6011],  
    [0.4424, 0.7783, 0.3338, 0.6097]
])

# Set style
sns.set(style="whitegrid")

# Create figure with subplots (3 pairs, 1 row)
fig, axes = plt.subplots(1, 2, figsize=(12, 5), sharey=True)

# Define colors for each metric
metric_colors = {'Precision': '#1f77b4', 'Normalized Recall': '#2ca02c', 
                 'MRR': '#d62728', 'Binary nDCG': '#9467bd'}
width = 0.3  # Bar width

# Iterate through each pair
for i, ax in enumerate(axes):
    x = np.arange(len(metrics))  # X positions for metrics
    values_bge = grouped_values[i * 2]  # BGE values
    values_reranker = grouped_values[i * 2 + 1]  # Trained Reranker values

    # Plot bars with distinct colors for each metric
    for j, metric in enumerate(metrics):
        ax.bar(x[j] - width/2, values_bge[j], width, color=metric_colors[metric], 
               alpha=0.6, edgecolor='black', label=f"BGE - {metric}" if i == 0 else "")
        ax.bar(x[j] + width/2, values_reranker[j], width, hatch='x', color=metric_colors[metric], 
               alpha=0.6, edgecolor='black', label=f"Trained Reranker - {metric}" if i == 0 else "")

        # Add value labels, bold the larger one
        bge_val = values_bge[j] * 100
        reranker_val = values_reranker[j] * 100
        bold_bge = 'bold' if bge_val > reranker_val else 'normal'
        bold_reranker = 'bold' if reranker_val > bge_val else 'normal'

        ax.text(x[j] - width/2, values_bge[j] + 0.005, f'{bge_val:.1f}%', 
                ha='center', va='bottom', fontsize=10, fontweight=bold_bge, color='black')

        ax.text(x[j] + width/2, values_reranker[j] + 0.005, f'{reranker_val:.1f}%', 
                ha='center', va='bottom', fontsize=10, fontweight=bold_reranker, color='black')

    # Formatting
    ax.set_xticks(x)
    ax.set_xticklabels(metrics, rotation=15, fontsize=10)
    ax.set_title(pair_labels[i], fontsize=12, fontweight='bold')
    ax.set_ylim(0, 1)  # Set same y-axis range for all subplots
    ax.yaxis.grid(True, linestyle='--', alpha=0.7)

# Overall figure adjustments
fig.suptitle("Comparison of BGE vs. Trained Reranker Performance", fontsize=14, fontweight='bold')
axes[0].set_ylabel("Performance Score (%)", fontsize=12)

# Create a custom legend for the metric colors
metric_legend_handles = [plt.Line2D([0], [0], color=color, lw=8) for color in metric_colors.values()]
metric_legend = fig.legend(metric_legend_handles, metric_colors.keys(), 
                           loc='upper center', fontsize=12, frameon=False, ncol=4, bbox_to_anchor=(0.5, 1.05))

# Create an extra legend for BGE vs. Trained Reranker (black with/without hatch)
bge_legend = plt.Rectangle((0, 0), 1, 1, alpha=0.2, label="BGE")

reranker_legend = plt.Rectangle((0, 0), 1, 1, alpha=0.2, hatch='xxx', linewidth = 2, edgecolor = 'black', label="Trained Reranker")

fig.legend(handles=[bge_legend, reranker_legend], labels=["BGE", "Trained Reranker"], 
           loc='upper center', fontsize=12, frameon=False, ncol=2, bbox_to_anchor=(0.5, 1.1))


plt.tight_layout()
plt.show()
plt.savefig('bar-graph-comparison-appendix.pdf')
```
